```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')

import warnings
warnings. filterwarnings('ignore')

%matplotlib inline
```

```python
df = pd.read_csv('/content/Diwali Sales Data.csv', encoding ='unicode_escape')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.drop(['Status','unnamed1'],axis = 1,inplace=True)
```

```python
pd.isnull(df)
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df['Amount'] = df['Amount'].astype('int')
```

```python
df['Amount'].dtypes
```

```python
df.columns
```

```python
df.rename(columns={'Marital_Status':'Shadi'})
```

# **Exploratory Data Analysis**



## Gender

```python
plt.figure(figsize= (18,8), dpi = 200)
sns.countplot(x = 'Gender', data = df)
```

```python
df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
```

```python
sales_gen = df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
```

```python
plt.figure(figsize= (18,8), dpi = 200)
sns.barplot(x = 'Gender',y= 'Amount', data = sales_gen)
```

From above graphs we can see that most buyers are females and also purchasing
power of females are greater than males



**Age**

```python
plt.figure(figsize= (18,8), dpi = 200)
sns.countplot(data=df,x ='Age Group',hue ='Gender')
```

```python
sales_age = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
```

```python
plt.figure(figsize= (18,8), dpi = 200)
sns.barplot(x='Age Group',y='Amount',data = sales_age)
```

*From above graphs we can see that most of the buyers are of age group between 26-35 yrs female*

# **State**

```python
sales_state = df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
```

```python
sns.set(rc={'figure.figsize' :(15,5)})
sns.barplot(x='State',y='Orders',data = sales_state)
```

```python
sales_state = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)
```

```python
sns.set(rc={'figure.figsize' :(15,5)})
sns.barplot(x='State',y='Amount',data = sales_state)
```

From above graphs we can see that unexpectedly most of the orders are from Uttar Pradesh, Maharastra and Karnataka respectively but total sales/amount is from UP, Karnataka and then Maharastra.

```python
plt.figure(figsize= (20,10), dpi = 200)
sns.countplot(x ='Marital_Status', data =df)
```

```python
sales_state = df.groupby(['Marital_Status','Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
```

```python
sns.set(rc={'figure.figsize':(6,5)})
sns.barplot(data=sales_state,x='Marital_Status',y='Amount',hue='Gender')
```

From above graphs we can see that most of the buyers are married women and they have high purchasing power.

```python
plt.figure(figsize= (25,10), dpi = 200)
sns.countplot(data=df,x='Occupation')
```

```python
sales_state = df.groupby(['Occupation'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
```

```python
plt.figure(figsize= (25,10), dpi = 200)
sns.barplot(data=sales_state,x='Occupation',y='Amount')
```

From above graphs we can see that most of the buyers are working in IT,Healthcare sector and Aviation sector.
